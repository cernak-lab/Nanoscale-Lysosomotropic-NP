```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import AllChem, Descriptors
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt

data1 = pd.read_csv('combined_Enamine_libray_leelamine_amine_product_enumeration_20231129.csv')  # Replace with the actual file name for the first file
data1 = data1[data1['Dataset']!='File2']
# Load your data into Pandas DataFrames
data2 = pd.read_csv('DHA_amine_product_enumeration_20240104.csv')  # Replace with the actual file name for the second file
data3 = pd.read_csv('LLM_amine_product_enumeration_20240104.csv')  # Replace with the actual file name for the third file

# Add a label to each dataset to distinguish them after concatenation
data1['Dataset'] = 'File1'
data2['Dataset'] = 'File2'
data3['Dataset'] = 'File3'

# Combine the three datasets into one DataFrame
data = pd.concat([data1,data2, data3], ignore_index=True)

# Save the combined DataFrame to a new CSV file
#data.to_csv('combined_three_files_20240104.csv', index=False)

# Calculate Morgan fingerprints for each molecule
morgan_fps = []
for smiles in data['product_smiles']:
    molecule = Chem.MolFromSmiles(smiles)
    morgan_fp = AllChem.GetMorganFingerprintAsBitVect(molecule, 2)
    morgan_fps.append(morgan_fp)

# Convert the list of Morgan fingerprints to a NumPy array
morgan_fps = np.array(morgan_fps)

# Apply t-SNE to the Morgan fingerprints
tsne = TSNE(n_components=2, perplexity=10, random_state=20)
embeddings = tsne.fit_transform(morgan_fps)

# Create masks to identify rows corresponding to each dataset
is_data1 = data['Dataset'] == 'File1'
is_data2 = data['Dataset'] == 'File2'
is_data3 = data['Dataset'] == 'File3'
```

```python
# Set a fixed size for the plot
fig = plt.figure(figsize=(6, 6))

plt.rcParams['font.family'] = 'Arial'
plt.rcParams['axes.labelsize'] = 10        # Axis labels size
plt.rcParams['axes.labelweight'] = 'bold'  # Axis labels bold
plt.rcParams['legend.fontsize'] = 8        # Legend font size

# Plot the t-SNE embeddings with colors representing each dataset
plt.scatter(embeddings[is_data1, 0], embeddings[is_data1, 1], c='#C5C9C7', label='1 + acids (Enamine), 5 + amines (Enamine)',alpha=0.9, s=15)
plt.scatter(embeddings[is_data2, 0], embeddings[is_data2, 1], c='#6B4596FF', label='5 + amines',alpha=0.9, s=15)
plt.scatter(embeddings[is_data3, 0], embeddings[is_data3, 1], c='#F68F46FF', label='1 + acids',alpha=0.9, s=15)

#plt.colorbar(label='Color')
plt.xlabel('t-SNE 1')
plt.ylabel('t-SNE 2')

# plt.axis('equal')
# plt.gca().set_aspect("equal", adjustable="box")
#plt.title('Three Files Colored on t-SNE Plot')
plt.legend()
resolution_value = 600
plt.savefig("comparison of d2b and Enamine library_tSNE_20240104.png", format="png", dpi=resolution_value)
plt.show()
```
